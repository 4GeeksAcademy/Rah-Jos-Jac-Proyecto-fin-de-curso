# Normalización básica de municipios

## 1. Importación de librerías

```python
import pandas as pd
import re
import os
from datetime import datetime
```

## 2. Carga de datos

```python
# Cargar datos raw
ruta_raw = r'C:\00 - Proyecto Incendios Galicia 0.0\data\01 - raw\01 - municipios\01 - municipios.csv'
df = pd.read_csv(ruta_raw)

print(f"Datos cargados: {len(df)} registros")
print(f"Columnas: {list(df.columns)}")
```

## 3. Función de normalización

```python
def normalizar_texto(texto):
    """Normaliza texto: minúsculas, sin paréntesis, espacios limpios"""
    if pd.isnull(texto):
        return ''
    texto = str(texto).strip().lower()
    texto = re.sub(r'\s*\(.*?\)', '', texto)  # Quitar paréntesis
    texto = texto.replace('provincia de ', '')  # Quitar prefijos
    texto = ' '.join(texto.split())  # Limpiar espacios
    return texto

print("Función de normalización definida")
```

## 4. Aplicar normalización

```python
# Normalizar columnas de texto
df['municipio'] = df['municipio'].apply(normalizar_texto)
df['comarca'] = df['comarca'].apply(normalizar_texto)
df['provincia'] = df['provincia'].apply(normalizar_texto)

# Ordenar alfabéticamente
df = df.sort_values('municipio').reset_index(drop=True)

print("Normalización aplicada")
print(f"Municipios únicos: {df['municipio'].nunique()}")
print(f"Provincias: {df['provincia'].unique()}")
```

## 5. Añadir municipios faltantes

```python
# Municipios faltantes identificados
municipios_adicionales = [
    {
        'municipio': 'cerceda',
        'comarca': 'comarca de órdenes',
        'provincia': 'a coruña',
        'altitud': '',
        'superficie': '111,27',
        'poblacion': '5019',
        'densidad': ''
    },
    {
        'municipio': 'carballo',
        'comarca': 'bergantinos',
        'provincia': 'a coruña',
        'altitud': '106',
        'superficie': '186,09',
        'poblacion': '',
        'densidad': ''
    }
]

# Añadir municipios
for municipio in municipios_adicionales:
    df = pd.concat([df, pd.DataFrame([municipio])], ignore_index=True)

# Reordenar
df = df.sort_values('municipio').reset_index(drop=True)

print(f"Municipios añadidos: {len(municipios_adicionales)}")
print(f"Total final: {len(df)} municipios")
```

## 6. Validación y estadísticas

```python
# Estadísticas de validación
print("ESTADÍSTICAS FINALES:")
print(f"   Total municipios: {len(df)}")
print(f"   Municipios únicos: {df['municipio'].nunique()}")

print("\nDISTRIBUCIÓN POR PROVINCIAS:")
conteo_provincias = df['provincia'].value_counts()
for provincia, cantidad in conteo_provincias.items():
    print(f"   {provincia}: {cantidad} municipios")

# Mostrar muestra
print("\nMUESTRA (primeros 5):")
display(df[['municipio', 'provincia', 'poblacion']].head())
```

## 7. Guardar datos normalizados

```python
# Guardar archivo normalizado
ruta_output = r'C:\00 - Proyecto Incendios Galicia 0.0\data\02 - Municipio normalizado\01 - municipios'
os.makedirs(ruta_output, exist_ok=True)

archivo_normalizado = os.path.join(ruta_output, '01 - municipios normalizados.csv')
df.to_csv(archivo_normalizado, index=False, encoding='utf-8-sig')

print(f"Archivo normalizado guardado: {archivo_normalizado}")
print(f"Registros guardados: {len(df)}")
print(f"Normalización completada: {datetime.now().strftime('%Y-%m-%d %H:%M:%S')}")
```

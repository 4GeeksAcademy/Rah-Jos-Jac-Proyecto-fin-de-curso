# Normalización completa de municipios

## 1 - Importacion librerias

```python
import pandas as pd
import os
from datetime import datetime
```

## 2 - Carga de datos normalizados

```python
# Cargar datos normalizados
ruta_input = r'C:\00 - Proyecto Incendios Galicia - END\data\02 - Municipio normalizado\01 - municipios\01 - municipios normalizados.csv'
df = pd.read_csv(ruta_input)

print(f"Datos cargados: {len(df)} municipios")
print(f"Columnas disponibles: {list(df.columns)}")
```

## 2. Validación final de datos

```python
# Validaciones críticas
print("VALIDACIONES FINALES:")

# Verificar municipios únicos
duplicados = df['municipio'].duplicated().sum()
print(f"   Municipios duplicados: {duplicados}")

# Verificar total esperado (315 municipios en Galicia)
total_esperado = 315
total_actual = len(df)
print(f"   Total esperado: {total_esperado}")
print(f"   Total actual: {total_actual}")
print(f"   Diferencia: {total_actual - total_esperado}")

# Verificar provincias
provincias_esperadas = {'a coruña', 'lugo', 'ourense', 'pontevedra'}
provincias_actuales = set(df['provincia'].unique())
print(f"   Provincias correctas: {provincias_esperadas == provincias_actuales}")

if duplicados > 0:
    print("ATENCIÓN: Hay municipios duplicados")
    display(df[df['municipio'].duplicated(keep=False)].sort_values('municipio'))
```

## 3. Optimización final del dataset

```python
# Seleccionar columnas esenciales para el archivo de referencia
columnas_finales = ['municipio', 'comarca', 'provincia']
df_final = df[columnas_finales].copy()

# Eliminr duplicados si existen
df_final = df_final.drop_duplicates(subset=['municipio']).reset_index(drop=True)

# Ordenar alfabéticamente
df_final = df_final.sort_values('municipio').reset_index(drop=True)

print(f"Dataset final optimizado: {len(df_final)} municipios únicos")
print(f"Columnas finales: {list(df_final.columns)}")
```

## 4. Generar archivos adicionales

```python
# Crear directorio de salida
ruta_output = r'C:\00 - Proyecto Incendios Galicia - END\data\03 - Normalizado completo\01 - municipios'
os.makedirs(ruta_output, exist_ok=True)

# Generar archivos .txt por provincia
print("Generando archivos por provincia:")
for provincia in df_final['provincia'].unique():
    municipios_provincia = df_final[df_final['provincia'] == provincia]['municipio'].sort_values()
    nombre_archivo = f"municipios_{provincia.replace(' ', '_').replace('á','a').replace('é','e').replace('í','i').replace('ó','o').replace('ú','u')}.txt"
    ruta_archivo = os.path.join(ruta_output, nombre_archivo)
    
    with open(ruta_archivo, 'w', encoding='utf-8') as f:
        for municipio in municipios_provincia:
            f.write(municipio + '\n')
    
    print(f"   {nombre_archivo}: {len(municipios_provincia)} municipios")
```

## 5. Guardar archivo de referencia final

```python
# Guardar archivo de referencia principal
archivo_referencia = os.path.join(ruta_output, '01 - municipios normalizado completo.csv')
df_final.to_csv(archivo_referencia, index=False, encoding='utf-8-sig')

print(f"ARCHIVO DE REFERENCIA FINAL GUARDADO:")
print(f"   Ruta: {archivo_referencia}")
print(f"   Municipios: {len(df_final)}")
print(f"   Columnas: {list(df_final.columns)}")

# Resumen final
print(f"\nDISTRIBUCIÓN FINAL POR PROVINCIAS:")
conteo_final = df_final['provincia'].value_counts().sort_index()
for provincia, cantidad in conteo_final.items():
    print(f"   {provincia.title()}: {cantidad} municipios")

print(f"\nNORMALIZACIÓN COMPLETA FINALIZADA")
print(f"Completado: {datetime.now().strftime('%Y-%m-%d %H:%M:%S')}")
print(f"Archivo de referencia listo para uso en el proyecto")
```

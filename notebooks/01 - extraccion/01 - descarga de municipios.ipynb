# Descarga de municipios de Galicia

## 1. Importación de librerías

```python
import pandas as pd
import requests
import os
from bs4 import BeautifulSoup
from datetime import datetime
```

## 2. Descarga de datos web

```python
# URL fuente
url = "https://15mpedia.org/wiki/Lista_de_municipios_de_Galicia"
print(f"Descargando desde: {url}")

try:
    response = requests.get(url)
    response.raise_for_status()
    soup = BeautifulSoup(response.content, 'html.parser')
    print("Página descargada correctamente")
except Exception as e:
    print(f"Error descargando: {e}")
    raise
```

## 3. Localización de tabla

```python
# Buscar tabla de municipios
tables = soup.find_all('table')
print(f"Tablas encontradas: {len(tables)}")

tabla_municipios = None
for i, tabla in enumerate(tables):
    filas = tabla.find_all('tr')
    print(f"   Tabla {i+1}: {len(filas)} filas")
    if len(filas) > 200:  # Tabla principal de municipios
        tabla_municipios = tabla
        print(f"Tabla de municipios identificada")
        break

if tabla_municipios is None:
    raise ValueError("No se encontró la tabla de municipios")
```

## 4. Extracción raw de datos

```python
# Extraer datos sin procesar
datos_raw = []
filas = tabla_municipios.find_all('tr')

for fila in filas:
    celdas = fila.find_all(['td', 'th'])
    if len(celdas) >= 6:
        fila_datos = [celda.get_text().strip() for celda in celdas]
        if 'Provincia de' in ' '.join(fila_datos):
            datos_raw.append(fila_datos)

print(f"Datos extraídos: {len(datos_raw)} registros")
```

## 5. Guardar datos

```python
# Crear DataFrame raw
columnas = ['municipio', 'comarca', 'provincia', 'altitud', 'superficie', 'poblacion', 'densidad']
df_raw = pd.DataFrame(datos_raw, columns=columnas)

# Guardar archivo raw
ruta_raw = r'C:\00 - Proyecto Incendios Galicia 0.0\data\01 - raw\01 - municipios'
os.makedirs(ruta_raw, exist_ok=True)

archivo_raw = os.path.join(ruta_raw, '01 - municipios.csv')
df_raw.to_csv(archivo_raw, index=False, encoding='utf-8-sig')

print(f"Archivo raw guardado: {archivo_raw}")
print(f"Registros guardados: {len(df_raw)}")
print(f"Extracción completada: {datetime.now().strftime('%Y-%m-%d %H:%M:%S')}")
```
